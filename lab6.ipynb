#EXP6: USE  MATPLOTLIB LIBRARY TO VISUALISE A LINE,SCATTER PLOT,BAR CHART,HISTOGRAM,PIE-CHART,BOX/WHISKER PLOT

```python
#draw a line in a diagram from position (1,3) to(8,10)
#x=(1,8)
#y=(3,10)
import matplotlib.pyplot as plt
import numpy as np
xpoints=np.array([1,8])
ypoints=np.array([3,10])
plt.plot(xpoints,ypoints)
plt.show()
```

```python
#draw a line with default x-axis starting with 0
ypoint=np.array([3,8,1,10,5,7])
#ypoints=data on the y axis
#marker=marker style
#ls=line style
#lw=line width
#mec=marker edge colour
#mfc=marker face colour(colour inside the marker)
plt.plot(ypoint,marker='o',ls='--',lw='2',ms=20,mec='r',mfc='y')
plt.show()
```

```python
#draw two lines by specifying a plt.plot() for each line
y1=np.array([3,8,1,10])
y2=np.array([6,2,7,11])
plt.plot(y1)
plt.plot(y2)
plt.show()
```

```python
#draw to lines by specifying the x and y point value for both lines
x1=np.array([0,1,2,3])
y1=np.array([3,8,1,10])
x2=np.array([0,1,2,3])
y2=np.array([6,2,7,11])
plt.plot(x1,y1,x2,y2)
plt.show()
```

```python
x=np.array([10,20,30,40,50,60,70,80])
y=np.array([100,120,130,140,150,160,170,180])
plt.plot(x,y,'o:r',mfc='b')
plt.title("health data of the a2 batch")
plt.xlabel("average pulse")
plt.ylabel("burnt calories")
plt.show()
#set font properties of the graph
font1={'family':'serif','color':'blue','size':20}
font2={'family':'serif','color':'darkred','size':15}
plt.plot(x,y,'o:r',mfc='b')
plt.title("health data of the a2 batch",fontdict=font1)
plt.xlabel("average pulse",fontdict=font2)
plt.ylabel("calories",fontdict=font2)
plt.grid()
```

```python
plt.grid(axis='x')
plt.grid(axis='y')
plt.grid(color='green',linestyle="--",linewidth=0.5)
plt.show()
```

```python
#display multiple plot on a sample plot
#subplot()---plot--->figure--->inside--->plot
plt.subplot(2,1,1)
plt.plot(x1,y1)
plt.subplot(2,1,2)
plt.plot(x2,y2)
plt.show()
```

```python
#scatter plot
import matplotlib.pyplot as plt
import numpy as np
x1=np.array([5,1,15,17,19,4,2,7,3,9,12,13,6])
y1=np.array([99,95,85,86,92,93,98,88,87,89,90,91,94])
#car data
#random-->numpy function to generate the random data
#randint-->numpy function to generate the random integer
#100-->random integer from 0 to 99
#size(13)-->the return array will be size of 13
color=np.random.randint(100,size=(13))
size=2*np.random.randint(100,size=(13))
plt.subplot(2,3,1)
plt.scatter(x1,y1,c=color,s=size,alpha=1,cmap='nipy_spectral')
```

```python
#bar chart
import matplotlib.pyplot as plt
import numpy as np
x=np.array(["A","B","C","D"])
y=np.array([3,8,1,10])
plt.subplot(2,1,1)
plt.bar(x,y,width=0.1)
plt.subplot(2,1,2)
plt.barh(x,y,color="red",height=0.5)
plt.show()
```

```python
#histogram
import numpy as np
#170:value should be concentrated at 170
#10:standard deviation
#250:no.of values
x=np.random.normal(170,10,250)
plt.hist(x)
plt.title("height histogram")
plt.xlabel("height of a person in cm")
plt.ylabel("number of people")
plt.show()
```

```python
#pie chart
y=np.array([35,25,25,15])
mylabels=["Apple","Banana","Grapes","Dragon_Fruit"]
plt.subplot(3,2,1)
plt.pie(y,labels=mylabels)
plt.subplot(3,2,2)
plt.pie(y,labels=mylabels,startangle=90)
plt.subplot(3,2,3)
myexplode=[0.2,0,0,0]
plt.pie(y,labels=mylabels,explode=myexplode)
plt.subplot(3,2,4)
plt.pie(y,labels=mylabels,explode=myexplode,shadow=True)
plt.subplot(3,2,5)
mycolor={"black","red","b"}
plt.pie(y,labels=mylabels,colors=mycolor)
plt.subplot(3,2,6)
plt.pie(y,labels=mylabels)
plt.legend(title='Four Fruits')
plt.show()
```

```python
#box plot
np.random.seed(10)
data_1=np.random.normal(100,10,200)
data_2=np.random.normal(90,20,200)
data_3=np.random.normal(80,30,200)
data_4=np.random.normal(70,40,200)
data=[data_1,data_2,data_3,data_4]
fig=plt.figure(figsize=(10,7))
#axes
ax=fig.add_axes([0,0,1,1])
#create the plot
bp=ax.boxplot(data)
plt.show()
```
