```python
#3. bar plot
import seaborn as sns
import matplotlib.pyplot as plt
categories=["A","B","C","D"]
values=[20,35,30,40]
sns.barplot(x=categories,y=values)
plt.show()
```

```python
#7. Box plot
import seaborn as sns
import matplotlib.pyplot as plt
data=[10,12,15,14,16,18,20,22,25,50]
sns.boxplot(y=data)
plt.show()
```

```python
#8.violin plot
import seaborn as sns
import matplotlib.pyplot as plt
data=[10,12,15,14,16,18,20,22,25,50]
sns.violinplot(y=data)
plt.show()
```

```python
#9. kde plot
import seaborn as sns
import matplotlib.pyplot as plt
data=[10,12,15,14,16,18,20,22,25,50]
sns.kdeplot(data=data)
plt.show()
```

```python
#10.heatmap
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
data={
    "Math":[80,90,70,85],
    "science":[85,95,75,80],
    "English":[70,80,90,75]
}
df=pd.DataFrame(data)
correlation=df.corr()
sns.heatmap(correlation, annot=True)
plt.show()
```
