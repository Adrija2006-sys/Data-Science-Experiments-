```python
#reading the csv file
import pandas as pd 
from pandas import read_csv
#plotting the graph
from matplotlib import pyplot
#Train-test split
from sklearn.model_selection import train_test_split
#Validation
from sklearn.model_selection import cross_val_score
#K-Fold
from sklearn.model_selection import StratifiedKFold
#importing logistic regression--->sklearn
from sklearn.linear_model import LogisticRegression
#importing KNN
from sklearn.neighbors import KNeighborsClassifier
```

```python
path="C:\\Users\\adrik\\Downloads\\iris.csv"
names=['sepal_length','sepal_width','petal_length','petal_width','class']
dataset=read_csv(path,names=names)
#print(dataset)
#split the dataset into training, testing,and validation
array=dataset.values
x=array[:,0:4]
y=array[:,4]
#we are creating a training and validation set--->using--->train_test_split()
#test_size=0.28---.20% percent of data will be used as a test data
#random state=1
#shuffle--->shuffling the data
x_train, x_validation, y_train, y_validation=train_test_split(x,y,test_size=0.20,random_state=1,shuffle=True)
```

```python
#Empty list for the models
models=[]
#models.append(('LR',LogisticRegression(solver='liblinear')))
models.append(('LR',LogisticRegression()))
models.append(('KNN',KNeighborsClassifier()))
print(models)
result=[]
names=[]
for name,model in models:
    kfold=StratifiedKFold(n_splits=10,random_state=1,shuffle=True)
    cv_result=cross_val_score(model,x_train,y_train,cv=kfold,scoring='accuracy')
    result.append(cv_result)
    names.append(name)
    print('%s:%f(%f)'%(name,cv_result.mean(),cv_result.std()))
#compare the algorithm
pyplot.boxplot(result,label=names)
pyplot.title('Algorithm Comparison')
pyplot.show()
```
