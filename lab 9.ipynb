```python
# drawing the chart(line chart)
import sys
import matplotlib.pyplot as plt
from scipy import stats
```

```python
# LINEAR REGRESSION=Y=Mx+C---->M(slope), C(intercept) ,x(ind.var)
x=[5,7,8,7,2,17,2,9,4,11,12,9,6]
y=[99,86,87,88,111,86,103,87,94,78,77,85,86]
#x and y are historical data (if x is 5 then y is 99)
```

```python
slope,intercept,r,p,std_err=stats.linregress(x,y)
def myfun(x):
    return slope*x+intercept #MX+C
z=map(myfun,x)
print(z)
# map function executes a specified function for each item in an iterable

mymodel=list(z) #convert the map into a list
print(mymodel)
```

```python
plt.scatter(x,y)
plt.plot(x,mymodel)
plt.show()
```
